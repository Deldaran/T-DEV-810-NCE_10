# MNIST dataset

## Download
We load the dataset directly from keras.

```python
from dotenv import load_dotenv
load_dotenv()

from keras.datasets import mnist

(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

from prettytable import PrettyTable

table = PrettyTable()
table.field_names = ["Dataset", "Shape", "Length"]

table.add_row(["train_images", train_images.shape, len(train_images)])
table.add_row(["train_labels", train_labels.shape, len(train_labels)])
table.add_row(["test_images", test_images.shape, len(test_images)])
table.add_row(["test_labels", test_labels.shape, len(test_labels)])

print(table)
```
